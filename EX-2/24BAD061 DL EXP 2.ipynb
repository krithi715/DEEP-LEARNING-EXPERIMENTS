Task B: PERFORMANCE COMPARISON OF ACTIVATION FUNCTIONS

```python
print("KRITHIKA K - 24BAD061")
import tensorflow as tf
import matplotlib.pyplot as plt
```

```python
(x_train,y_train),(x_test,y_test)=tf.keras.datasets.mnist.load_data()
x_train=x_train/255.0
x_test=x_test/255.0
```

```python
def create_model(activation):
    model = tf.keras.Sequential([
        tf.keras.layers.Flatten(input_shape=(28, 28)),
        tf.keras.layers.Dense(128, activation=activation),
        tf.keras.layers.Dense(10, activation="softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model
```

```python
#train with sigmoid
sigmoid_model = create_model("sigmoid")
sigmoid_history = sigmoid_model.fit(
    x_train,
    y_train,
    epochs=20,
    validation_split=0.2,
    verbose=1
)
```

```python
#train with tanh
tanh_model = create_model("tanh")
tanh_history= tanh_model.fit(
    x_train,y_train,
    epochs=20,
    validation_split=0.2,
    verbose=1
)
```

```python
#train with ReLU
relu_model= create_model('relu')
relu_history= relu_model.fit(
    x_train,y_train,
    epochs=20,
    validation_split=0.2,
    verbose=1
)
```

```python
plt.plot(sigmoid_history.history['val_accuracy'],label="Sigmoid")
plt.plot(tanh_history.history['val_accuracy'],label="Tanh")
plt.plot(relu_history.history['val_accuracy'],label="ReLU")
plt.title("Validation Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()
plt.show()
```

```python
plt.plot(sigmoid_history.history['val_loss'],label="Sigmoid")
plt.plot(tanh_history.history['val_loss'],label="Tanh")
plt.plot(relu_history.history['val_loss'],label="ReLU")
plt.title("Validation loss")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()
plt.show()
```

```python
print("Sigmoid")
print("Training Accuracy:", sigmoid_history.history['accuracy'][-1])
print("Validation Accuracy:", sigmoid_history.history['val_accuracy'][-1])
print("Training Loss:", sigmoid_history.history['loss'][-1])
print("Validation Loss:", sigmoid_history.history['val_loss'][-1])

print("\nTanh")
print("Training Accuracy:", tanh_history.history['accuracy'][-1])
print("Validation Accuracy:", tanh_history.history['val_accuracy'][-1])
print("Training Loss:", tanh_history.history['loss'][-1])
print("Validation Loss:", tanh_history.history['val_loss'][-1])

print("\nReLU")
print("Training Accuracy:", relu_history.history['accuracy'][-1])
print("Validation Accuracy:", relu_history.history['val_accuracy'][-1])
print("Training Loss:", relu_history.history['loss'][-1])
print("Validation Loss:", relu_history.history['val_loss'][-1])
```

Task C: COMPARISON OF OPTIMIZATION ALGORITHMS

```python
print("KRIHTIKA K - 24BAD061")
import tensorflow as tf
import matplotlib.pyplot as plt
```

```python
def create_model(optimizer):
    model=tf.keras.Sequential([
        tf.keras.layers.Flatten(input_shape=(28,28)),
        tf.keras.layers.Dense(128,activation="relu"),
        tf.keras.layers.Dense(10,activation="softmax")
    ])
    model.compile(
        optimizer=optimizer,
        loss="sparse_categorical_crossentropy",
        metrics=['accuracy']
    )
    return model
```

```python
#optimizers
sgd=tf.keras.optimizers.SGD(learning_rate=0.01)

momentum=tf.keras.optimizers.SGD(
    learning_rate=0.01,
    momentum=0.9
)
rmsprop=tf.keras.optimizers.RMSprop(learning_rate=0.001)
adam=tf.keras.optimizers.Adam(learning_rate=0.001)
```

```python
#model training
model_sgd=create_model(sgd)
history_sgd =model_sgd.fit(
    x_train,y_train,
    epochs=10,
    validation_split=0.2,
    verbose=1
)

model_momentum=create_model(momentum)
history_momentum =model_momentum.fit(
    x_train,y_train,
    epochs=10,
    validation_split=0.2,
    verbose=1
)

model_rmsprop=create_model(rmsprop)
history_rmsprop =model_rmsprop.fit(
    x_train,y_train,
    epochs=10,
    validation_split=0.2,
    verbose=1
)

model_adam=create_model(adam)
history_adam =model_adam.fit(
    x_train,y_train,
    epochs=10,
    validation_split=0.2,
    verbose=1
)
```

```python
plt.plot(history_sgd.history["loss"], label="SGD")
plt.plot(history_momentum.history["loss"],label="Momentum")
plt.plot(history_rmsprop.history["loss"],label="RMSprop")
plt.plot(history_adam.history["loss"],label="Adam")
plt.title("Training Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Training loss")
plt.grid()
plt.legend()
plt.show
```

```python
plt.plot(history_sgd.history["accuracy"], label="SGD")
plt.plot(history_momentum.history["accuracy"],label="Momentum")
plt.plot(history_rmsprop.history["accuracy"],label="RMSprop")
plt.plot(history_adam.history["accuracy"],label="Adam")

plt.title("Training Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Training Accuracy")
plt.grid()
plt.legend()
plt.show
```

```python
print("SGD")
print("Training Accuracy:", history_sgd.history["accuracy"][-1])
print("Validation Accuracy:", history_sgd.history["val_accuracy"][-1])
print("Training Loss:", history_sgd.history["loss"][-1])
print("Validation Loss:", history_sgd.history["val_loss"][-1])

print("\nMomentum")
print("Training Accuracy:", history_momentum.history["accuracy"][-1])
print("Validation Accuracy:", history_momentum.history["val_accuracy"][-1])
print("Training Loss:", history_momentum.history["loss"][-1])
print("Validation Loss:", history_momentum.history["val_loss"][-1])

print("\nRMSProp")
print("Training Accuracy:", history_rmsprop.history["accuracy"][-1])
print("Validation Accuracy:", history_rmsprop.history["val_accuracy"][-1])
print("Training Loss:", history_rmsprop.history["loss"][-1])
print("Validation Loss:", history_rmsprop.history["val_loss"][-1])

print("\nAdam")
print("Training Accuracy:", history_adam.history["accuracy"][-1])
print("Validation Accuracy:", history_adam.history["val_accuracy"][-1])
print("Training Loss:", history_adam.history["loss"][-1])
print("Validation Loss:", history_adam.history["val_loss"][-1])
```
